```python
import pandas as pd
import joblib
from sklearn.metrics.pairwise import cosine_similarity

# Load the catalog and the saved matrix
df = pd.read_csv('../data/product_catalog.csv')
tfidf_matrix = joblib.load('../data/vectorized_features.pkl')

# Calculate the similarity between ALL products
# This creates a score from 0 (nothing in common) to 1 (identical)
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

print("Similarity scores calculated.")
```

```python
def get_recommendations(product_name, top_n=3):
    # Check if the product exists (case-insensitive)
    matches = df[df['product_name'].str.lower() == product_name.lower()]

    if matches.empty:
        print(f"Product '{product_name}' not found in catalog!")
        print(f"Try one of these: {df['product_name'].head(3).tolist()}")
        return

    # Get the index
    idx = matches.index[0]

    # Get similarity scores
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # Get top matches
    top_matches = sim_scores[1:top_n+1]

    print(f"If you like '{df['product_name'].iloc[idx]}', you might also like:")
    for i, score in top_matches:
        print(f"- {df['product_name'].iloc[i]} (Match Score: {score:.2f})")
```
