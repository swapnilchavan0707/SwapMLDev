```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
import joblib
import os

# Load the catalog we generated
df = pd.read_csv('../data/product_catalog.csv')
print(f"Loaded {len(df)} products.")
df.head()
```

```python
# Initialize the Vectorizer
# stop_words='english' removes common words like 'and', 'the', etc.
tfidf = TfidfVectorizer(stop_words='english')

# Transform the description_tags into a matrix of numbers
tfidf_matrix = tfidf.fit_transform(df['description_tags'])

print(f"Matrix Shape: {tfidf_matrix.shape}")
print("The AI now sees the data as a grid of numbers!")

# Save the vectorizer and the matrix for the dashboard
os.makedirs('../models', exist_ok=True)
joblib.dump(tfidf, '../models/tfidf_transformer.pkl')
joblib.dump(tfidf_matrix, '../data/vectorized_features.pkl')
print("Vectorization complete and saved.")
```
