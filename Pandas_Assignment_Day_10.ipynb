## Applying functions to columns and creating new columns

We need to create another column in d3, which contains  a boolean value for each city to indicate whether it's a union territory or not.
- HINT: Chandigarh, Pondicherry and Delhi are only 3 union territories here.


```python
# write function here

# def is_ut(x):

#     # write code below

# d3['is_ut'] =

import pandas as pd

# Recreate DataFrame 'df' to its current state
points_column_data = [100, 200, 300, 400, 800, 450, 100]
common_index_df = ['c', 'a', 'b', 'd', 'e', 'f', 'g']
fruits_column_data = ['mango', 'apple', 'apple', 'apple', 'orange', 'mango', 'mango']
df = pd.DataFrame({
    'points': points_column_data
}, index=common_index_df)
df['fruits'] = pd.Series(fruits_column_data, index=common_index_df)

# Recreate DataFrame 'd1' to its current state
data_d1 = {
    'city': ['Chandigarh', 'Delhi', 'Kanpur', 'Chennai', 'Manali'],
    'Temperature': [15, 22, 20, 26, -2]
}
d1 = pd.DataFrame(data_d1)

# Merge (concatenate side-by-side) the two dataframes, df and d1, to get d3
d3 = pd.concat([df, d1], axis=1)

# Define the list of Union Territories
union_territories_list = ['Chandigarh', 'Delhi', 'Pondicherry']

# Define the function to check if a city is a Union Territory
def is_ut(city_name):
    # Check if the city_name is in the list of union territories
    # Handle NaN values gracefully by returning False if city_name is NaN
    if pd.isna(city_name):
        return False
    return city_name in union_territories_list

# Apply the function to the 'city' column to create the 'is_ut' column
d3['is_ut'] = d3['city'].apply(is_ut)
```

```python
# write function here
import pandas as pd

# Recreate DataFrame 'df' to its current state
points_column_data = [100, 200, 300, 400, 800, 450, 100]
common_index_df = ['c', 'a', 'b', 'd', 'e', 'f', 'g']
fruits_column_data = ['mango', 'apple', 'apple', 'apple', 'orange', 'mango', 'mango']
df = pd.DataFrame({
    'points': points_column_data
}, index=common_index_df)
df['fruits'] = pd.Series(fruits_column_data, index=common_index_df)

# Recreate DataFrame 'd1' to its current state
data_d1 = {
    'city': ['Chandigarh', 'Delhi', 'Kanpur', 'Chennai', 'Manali'],
    'Temperature': [15, 22, 20, 26, -2]
}
d1 = pd.DataFrame(data_d1)

# Merge (concatenate side-by-side) the two dataframes, df and d1, to get d3
d3 = pd.concat([df, d1], axis=1)

# Define the list of Union Territories (for 'is_ut' column recreation)
union_territories_list = ['Chandigarh', 'Delhi', 'Pondicherry']

# Define the function to check if a city is a Union Territory (for 'is_ut' column recreation)
def is_ut(city_name):
    if pd.isna(city_name):
        return False
    return city_name in union_territories_list

# Recreate 'is_ut' column in d3
d3['is_ut'] = d3['city'].apply(is_ut)

# Define the function to convert Celsius to Fahrenheit
def c_to_f(temp_celsius):
    # Handle NaN values gracefully
    if pd.isna(temp_celsius):
        return float('NaN')
    return (temp_celsius * 9/5) + 32

# Create the new column 'Temperature_Fahrenheit'
d3['Temperature_Fahrenheit'] = d3['Temperature'].apply(c_to_f)

# Print the updated d3 DataFrame (or its relevant parts for verification)
print(d3)




```

```python
# Use Groupby of single column with aggregate sum()
import pandas as pd

# Recreate the DataFrame df_choice
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Laptop', 'Mouse'],
    'Price_USD': [1200, 25, 75, 300, 50, 1100, 30],
    'Quantity_Sold': [50, 200, 150, 30, 80, 70, 100] # Added more data for meaningful sum
}
df_choice = pd.DataFrame(data)

# Use Groupby on the 'Product' column and aggregate with sum() on 'Quantity_Sold'
# I am choosing 'Product' as the grouping column and 'Quantity_Sold' as the column to sum.
grouped_sum = df_choice.groupby('Product')['Quantity_Sold'].sum()

print("Groupby 'Product' and sum 'Quantity_Sold':")
print(grouped_sum)

# or

import pandas as pd

# Recreate the DataFrame df_choice (with some added data for better demonstration)
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Laptop', 'Mouse'],
    'Price_USD': [1200, 25, 75, 300, 50, 1100, 30],
    'Quantity_Sold': [50, 200, 150, 30, 80, 70, 100]
}
df_choice = pd.DataFrame(data)

# Group by 'Product' and sum 'Quantity_Sold'
grouped_sum = df_choice.groupby('Product')['Quantity_Sold'].sum()

print("Groupby 'Product' and sum 'Quantity_Sold':")
print(grouped_sum)

```

```python
# Use Groupby of single column with aggregate count()
import pandas as pd

# Recreate the DataFrame df_choice
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Laptop', 'Mouse'],
    'Price_USD': [1200, 25, 75, 300, 50, 1100, 30],
    'Quantity_Sold': [50, 200, 150, 30, 80, 70, 100]
}
df_choice = pd.DataFrame(data)

# Group by 'Product' and count 'Quantity_Sold'
grouped_count = df_choice.groupby('Product')['Quantity_Sold'].count()

print("Groupby 'Product' and count 'Quantity_Sold':")
print(grouped_count)
```

```python
# Use Groupby of single column with aggregate min() and max()
import pandas as pd

# Recreate the DataFrame df_choice
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Laptop', 'Mouse'],
    'Price_USD': [1200, 25, 75, 300, 50, 1100, 30],
    'Quantity_Sold': [50, 200, 150, 30, 80, 70, 100]
}
df_choice = pd.DataFrame(data)

# Use Groupby on the 'Product' column and aggregate with min() and max() on 'Quantity_Sold'
grouped_min_max = df_choice.groupby('Product')['Quantity_Sold'].agg(['min', 'max'])

print("Groupby 'Product' and aggregate min() and max() of 'Quantity_Sold':")
print(grouped_min_max)

#or
import pandas as pd

# Recreate the DataFrame df_choice
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Laptop', 'Mouse'],
    'Price_USD': [1200, 25, 75, 300, 50, 1100, 30],
    'Quantity_Sold': [50, 200, 150, 30, 80, 70, 100]
}
df_choice = pd.DataFrame(data)

# Group by 'Product' and aggregate min() and max() of 'Quantity_Sold'
grouped_min_max = df_choice.groupby('Product')['Quantity_Sold'].agg(['min', 'max'])

print("Groupby 'Product' and aggregate min() and max() of 'Quantity_Sold':")
print(grouped_min_max)

```

```python
# Use Groupby of any 2 columns with aggregate mean()
import pandas as pd

# Recreate the DataFrame df_choice
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Laptop', 'Mouse'],
    'Price_USD': [1200, 25, 75, 300, 50, 1100, 30],
    'Quantity_Sold': [50, 200, 150, 30, 80, 70, 100]
}
df_choice = pd.DataFrame(data)

# Group by 'Product' and 'Price_USD' and aggregate mean() of 'Quantity_Sold'
grouped_mean = df_choice.groupby(['Product', 'Price_USD'])['Quantity_Sold'].mean()

print("Groupby 'Product' and 'Price_USD' and aggregate mean() of 'Quantity_Sold':")
print(grouped_mean)

```

```python
# Use Groupby of any 2 columns with aggregate min() and max()
import pandas as pd

# Recreate the DataFrame df_choice
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Laptop', 'Mouse'],
    'Price_USD': [1200, 25, 75, 300, 50, 1100, 30],
    'Quantity_Sold': [50, 200, 150, 30, 80, 70, 100]
}
df_choice = pd.DataFrame(data)

# Use Groupby on two columns ('Product' and 'Price_USD') and aggregate with min() and max() on 'Quantity_Sold'
grouped_min_max_two_cols = df_choice.groupby(['Product', 'Price_USD'])['Quantity_Sold'].agg(['min', 'max'])

print("Groupby 'Product' and 'Price_USD' and aggregate min() and max() of 'Quantity_Sold':")
print(grouped_min_max_two_cols)

```
