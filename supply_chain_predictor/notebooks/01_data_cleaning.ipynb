```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set visual style
sns.set_theme(style="whitegrid")
%matplotlib inline

# Path setup (moving up one folder from 'notebooks' to root)
DATA_PATH = os.path.join('..', 'data', 'sales_history.csv')

print("Libraries imported and paths set.")
```

```python
# Load the data
df = pd.read_csv(DATA_PATH)

# Convert date to datetime objects
df['date'] = pd.to_datetime(df['date'])

# Sorting is CRITICAL for time-series forecasting
df = df.sort_values(by='date').reset_index(drop=True)

print("Dataset Info:")
print(df.info())
print("\nFirst 5 rows:")
df.head()
```

```python
# Extract time features
df['day_of_week'] = df['date'].dt.dayofweek  # 0=Monday, 6=Sunday
df['is_weekend'] = df['day_of_week'].apply(lambda x: 1 if x >= 5 else 0)
df['month'] = df['date'].dt.month

print("Time features extracted.")
df[['date', 'day_of_week', 'is_weekend', 'month']].head()
```

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='date', y='units_sold', marker='o', color='teal')

# Highlight promotions to see if they explain the spikes
promo_days = df[df['is_promotion'] == 1]
plt.scatter(promo_days['date'], promo_days['units_sold'], color='red', label='Promotion Active', zorder=5)

plt.title("Units Sold Over Time (Trend Analysis)")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
corr = df[['price', 'is_promotion', 'weather_score', 'units_sold', 'is_weekend']].corr()
sns.heatmap(corr, annot=True, cmap='YlGnBu', fmt=".2f")
plt.title("Correlation Matrix: Demand Drivers")
plt.show()
```

```python
# Ensure the directory exists
os.makedirs('../data', exist_ok=True)

# Save to CSV
processed_path = '../data/sales_history_cleaned.csv'
df.to_csv(processed_path, index=False)

print(f"Success! Cleaned data saved to: {processed_path}")
```
