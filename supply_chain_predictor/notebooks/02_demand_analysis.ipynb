```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Visual settings
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

# Load the cleaned data
DATA_PATH = os.path.join('..', 'data', 'sales_history_cleaned.csv')
df = pd.read_csv(DATA_PATH)
df['date'] = pd.to_datetime(df['date'])

print("Cleaned data loaded for analysis.")
```

```python
# Group by day name
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df['day_name'] = df['date'].dt.day_name()

plt.figure(figsize=(10, 5))
sns.barplot(x='day_name', y='units_sold', data=df, order=day_order, palette='viridis', hue='day_name', legend=False)
plt.title("Average Units Sold by Day of the Week")
plt.ylabel("Avg Units Sold")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.regplot(x='price', y='units_sold', data=df, scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title("Price vs. Units Sold (Price Elasticity)")
plt.xlabel("Price ($)")
plt.ylabel("Units Sold")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.violinplot(x='weather_score', y='units_sold', data=df, inner="quart", palette="Pastel1")
plt.title("Distribution of Sales by Weather Score")
plt.xlabel("Weather Score (Higher = Better Weather)")
plt.show()
```

```python
from pandas.plotting import autocorrelation_plot

plt.figure(figsize=(10, 5))
autocorrelation_plot(df['units_sold'])
plt.title("Autocorrelation of Units Sold")
plt.show()
```
