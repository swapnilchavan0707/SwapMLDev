```python
import pandas as pd
import numpy as np

# Load the dataset
# Ensure the filename matches the one in your environment
file_name = 'Delinquency_prediction_dataset.xlsx'
df = pd.read_excel(file_name)

# ==========================================
# STEP 1: REVIEW DATASET & IDENTIFY INSIGHTS
# ==========================================
print("--- STEP 1: INITIAL DATA AUDIT ---")

# 1.1 Basic Stats
print(f"Dataset Dimensions: {df.shape}")
print("\nMissing Values per Column:")
print(df.isnull().sum())

# 1.2 Identify Inconsistencies in Categorical Data
print("\nUnique Values in Employment_Status (Spotting Inconsistencies):")
print(df['Employment_Status'].unique())

# 1.3 Identify Potential Outliers or Anomalies
print("\nSummary Statistics for Numerical Columns:")
print(df.describe())

# ==========================================
# STEP 2: ADDRESS MISSING DATA & QUALITY
# ==========================================
print("\n--- STEP 2: CLEANING & IMPUTATION ---")

# 2.1 Impute Income with Median (Robust to outliers)
income_median = df['Income'].median()
df['Income'] = df['Income'].fillna(income_median)
print(f"Imputed Income with Median: {income_median}")

# 2.2 Impute Loan_Balance with Mean
loan_mean = df['Loan_Balance'].mean()
df['Loan_Balance'] = df['Loan_Balance'].fillna(loan_mean)
print(f"Imputed Loan_Balance with Mean: {loan_mean}")

# 2.3 Impute Credit_Score with Median
cs_median = df['Credit_Score'].median()
df['Credit_Score'] = df['Credit_Score'].fillna(cs_median)

# 2.4 Standardize Employment_Status
# Mapping various strings like 'EMP' and 'employed' to 'Employed'
def standardize_employment(status):
    if pd.isna(status): return "Unknown"
    status = str(status).strip().lower()
    if status in ['emp', 'employed']:
        return 'Employed'
    return status.capitalize()

df['Employment_Status'] = df['Employment_Status'].apply(standardize_employment)
print("Standardized Employment_Status column.")

# ==========================================
# STEP 3: DETECT PATTERNS & RISK FACTORS
# ==========================================
print("\n--- STEP 3: RISK FACTOR ANALYSIS ---")

# 3.1 Calculate Correlation with Delinquency
# Selecting only numerical columns for correlation
numerical_cols = df.select_dtypes(include=[np.number])
correlation_matrix = numerical_cols.corr()
risk_correlations = correlation_matrix['Delinquent_Account'].sort_values(ascending=False)

print("Correlation of variables with Delinquent_Account:")
print(risk_correlations)

# 3.2 Compare Risk Factors between Delinquent vs Non-Delinquent groups
risk_comparison = df.groupby('Delinquent_Account')[['Credit_Utilization', 'Debt_to_Income_Ratio', 'Missed_Payments']].mean()
print("\nMean Risk Factors by Delinquency Status:")
print(risk_comparison)

# ==========================================
# STEP 4: SAVE CLEANED DATA
# ==========================================
cleaned_file = 'Geldium_Cleaned_Dataset.csv'
df.to_csv(cleaned_file, index=False)
print(f"\nTask Complete! Cleaned dataset saved as: {cleaned_file}")
```
