```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import os

# Create data directory if it doesn't exist
os.makedirs('../data', exist_ok=True)
```

```python
def generate_iot_stream(points=50):
    start_time = datetime.now()
    data = []

    # Starting in NYC
    lat, lon = 40.7128, -74.0060
    temp = 2.0  # Goal is to stay near 2°C (Cold Chain)

    for i in range(points):
        # Move slightly Northwest
        lat += 0.01 + np.random.uniform(-0.002, 0.002)
        lon -= 0.015 + np.random.uniform(-0.002, 0.002)

        # Simulate Temperature with a spike (Anomaly)
        if 20 < i < 30:
            temp += np.random.uniform(0.5, 1.5)  # Rising temperature
        else:
            temp = 2.0 + np.random.uniform(-0.3, 0.3)

        data.append({
            "timestamp": start_time + timedelta(minutes=i*10),
            "shipment_id": "SHIP_101",
            "latitude": lat,
            "longitude": lon,
            "temperature": round(temp, 2),
            "humidity": round(80 + np.random.uniform(-5, 5), 2),
            "vibration": round(np.random.uniform(0.01, 0.5), 3)
        })

    return pd.DataFrame(data)

df_stream = generate_iot_stream()
df_stream.to_csv('../data/raw_sensor_stream.csv', index=False)
print("Simulated IoT stream saved to data/raw_sensor_stream.csv")
df_stream.head()
```
